# MS-RCPSP Benchmark Report

Notebook này dùng để đọc kết quả benchmark có sẵn hoặc chạy lại 4 thuật toán chính trên Google Colab.

```python
from pathlib import Path
import subprocess
import sys

RUN_FULL_BENCHMARK = False
ROOT = Path.cwd()
RESULTS_DIR = ROOT / 'results' / 'final_benchmark'
FIGURES_DIR = ROOT / 'figures'
REPORT_ASSETS_DIR = ROOT / 'report_assets'
FIGURES_DIR.mkdir(exist_ok=True)
REPORT_ASSETS_DIR.mkdir(exist_ok=True)

if (ROOT / 'requirements.txt').exists():
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_rows', 200)
pd.set_option('display.max_columns', 50)
```

## 1. Dataset Manifest

```python
manifest = pd.read_csv(ROOT / 'datasets' / 'manifest.csv')
manifest
```

## 2. Main Result Table

```python
main_table = pd.read_csv(RESULTS_DIR / 'main_4_algorithms_total_score_runtime_table.csv')
main_table
```

## 3. Optional Full Benchmark

Mặc định notebook chỉ đọc kết quả có sẵn. Đổi `RUN_FULL_BENCHMARK = True` ở cell đầu nếu muốn chạy lại toàn bộ.

```python
if RUN_FULL_BENCHMARK:
    command = [
        sys.executable,
        'src/benchmark_algorithms.py',
        '--algorithms', 'hs,ga,hybrid_hs_ga,hybrid_hs_ga_guided_mutation',
        '--seeds', '1..30',
        '--max-evaluations', '1000',
        '--fixed-budget',
        '--harmony-memory-size', '30',
        '--population-size', '50',
        '--hybrid-hs-ratios', '0.15',
        '--output-dir', 'results/benchmark_runs',
    ]
    subprocess.run(command, check=True)
    benchmark_dir = ROOT / 'results' / 'benchmark_runs'
else:
    benchmark_dir = RESULTS_DIR

benchmark_dir
```

## 4. Summary Results
